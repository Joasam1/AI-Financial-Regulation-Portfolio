# Exercise 02: Bank Risk Classification and Supervisory Analytics

Four-state classification using illustrative CAR and NPL thresholds. This is a clean, executable portfolio version of the exercise.

```python
import math

banks = [
    {"name": "Bank A", "car": 18, "npl_ratio": 8},
    {"name": "Bank B", "car": 14, "npl_ratio": 13},
    {"name": "Bank C", "car": 11, "npl_ratio": 7},
    {"name": "Bank D", "car": 10, "npl_ratio": 16},
    {"name": "Bank E", "car": 13, "npl_ratio": 10},
]
```

## 1. Explore the bank records

```python
print(banks)
for bank in banks:
    print(bank["name"])
print("Number of banks:", len(banks))
name = banks[2]["name"]
car = banks[2]["car"]
print(name, car)
```

## 2. Validate inputs

```python
def validate_bank(bank):
    car = bank.get("car")
    npl = bank.get("npl_ratio")

    # Check for missing values
    if car is None or npl is None:
        return False

    # Check CAR data type
    if not isinstance(car, (int, float)):
        return False

    # Check NPL data type
    if not isinstance(npl, (int, float)):
        return False

    # Check boolean data type
    if isinstance(car, bool) or isinstance(npl, bool):
        return False

    # Check for non-finite values
    if not math.isfinite(car) or not math.isfinite(npl):
        return False

    # Check numerical ranges
    if car < 0 or npl < 0 or npl > 100:
        return False
    else:
        return True
```

```python
print(validate_bank(banks[0]))
test_bank = {"name": "Test Bank", "car": -5, "npl_ratio": 12}
print(validate_bank(test_bank))

test_banks = [
    {"name": "Test A", "car": 15, "npl_ratio": 8},
    {"name": "Test B", "car": -2, "npl_ratio": 10},
    {"name": "Test C", "car": 14, "npl_ratio": -5},
    {"name": "Test D", "car": 16, "npl_ratio": 120},
    {"name": "Test E", "car": 13, "npl_ratio": 100},
]
for bank in test_banks:
    print(bank["name"], validate_bank(bank))
```

```python
test_banks = [
    {"name": "Bank A", "car": 15, "npl_ratio": 8},
    {"name": "Bank B", "car": True, "npl_ratio": 12},
    {"name": "Bank C", "car": 14, "npl_ratio": False},
    {"name": "Bank D", "car": "15", "npl_ratio": 10},
    {"name": "Bank E", "car": 12, "npl_ratio": -3},
    {"name": "Bank F", "car": 13, "npl_ratio": 100},
]
for bank in test_banks:
    print(bank["name"], validate_bank(bank))
```

## 3. Classify banks into four supervisory risk states

```python
def classify_bank(bank):
    # Validate first to prevent missing-key errors
    if not validate_bank(bank):
        return "Invalid Data"

    car = bank["car"]
    npl = bank["npl_ratio"]

    if car >= 13 and npl <= 10:
        return "Green — Stable"
    elif car >= 13 and npl > 10:
        return "Blue — Asset Quality Risk"
    elif car < 13 and npl <= 10:
        return "Yellow — Capital Risk"
    else:
        return "Red — Combined Risk"

for bank in banks:
    print(bank["name"], ":", classify_bank(bank))
```

## 4. Count risk categories and calculate percentages

```python
risk_counts = {"Green": 0, "Blue": 0, "Yellow": 0, "Red": 0}
invalid_count = 0
banks_with_invalid = banks + [{"name": "Bank F", "car": -5, "npl_ratio": 15}]

for bank in banks_with_invalid:
    status = classify_bank(bank)
    if status == "Invalid Data":
        invalid_count += 1
    elif status == "Green — Stable":
        risk_counts["Green"] += 1
    elif status == "Blue — Asset Quality Risk":
        risk_counts["Blue"] += 1
    elif status == "Yellow — Capital Risk":
        risk_counts["Yellow"] += 1
    elif status == "Red — Combined Risk":
        risk_counts["Red"] += 1

total_input_banks = len(banks_with_invalid)
total_classified_banks = sum(risk_counts.values())
print("Risk counts:", risk_counts)
print("Invalid records:", invalid_count)
print("Reconciliation successful:", total_input_banks == total_classified_banks + invalid_count)

print("SUPERVISORY RISK SUMMARY")
for category, count in risk_counts.items():
    if total_classified_banks > 0:
        print(category, count, (count / total_classified_banks) * 100)
    else:
        print(category, count, "Percentage unavailable")
print("Total classified banks:", total_classified_banks)
print("Invalid records:", invalid_count)
print("Total input banks:", total_input_banks)
```

## 5. Boundary, missing-data, and non-finite-value checks

```python
test_bank = {"name": "Bank G", "car": 13, "npl_ratio": 10}
print(validate_bank(test_bank))
print(classify_bank(test_bank))

test_banks = [
    {"name": "Bank H", "car": 15},
    {"name": "Bank I", "npl_ratio": 8},
    {"name": "Bank J", "car": None, "npl_ratio": 12},
    {"name": "Bank K", "car": 14, "npl_ratio": 7},
]
for bank in test_banks:
    print(bank["name"], validate_bank(bank), classify_bank(bank))

car = float("nan")
print(isinstance(car, float))
print(car < 0)
print(car >= 13)

for value in [15, float("nan"), float("inf"), float("-inf")]:
    print("Finite:", math.isfinite(value))
```
